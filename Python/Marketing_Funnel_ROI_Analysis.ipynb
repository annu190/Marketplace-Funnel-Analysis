2. Python Data Cleaning & Analysis

2.1. Which channel generates the most revenue?

```python
channel_revenue = (
    df.groupby("Channel")["Revenue (INR)"]
    .sum()
    .sort_values(ascending=False)
)

channel_revenue
```

2.2. Which Channel Has the Highest ROI?

```python
channel_roi = (
    df.groupby("Channel")
    .agg(
        Total_Spend=("Ad Spend (INR)", "sum"),
        Total_Revenue=("Revenue (INR)", "sum")
    )
)

channel_roi["ROI (%)"] = (
    (channel_roi["Total_Revenue"] - channel_roi["Total_Spend"])
    / channel_roi["Total_Spend"] * 100
).round(2)

channel_roi.sort_values("ROI (%)", ascending=False)
```

2.3. Which Channel Generates the Most Leads?

```python
channel_leads = (
    df.groupby("Channel")["Leads"]
    .sum()
    .sort_values(ascending=False)
)

channel_leads
```

2.4. Which Channel Has the Lowest CAC?

```python
channel_cac = (
    df.groupby("Channel")
    .agg(
        Spend=("Ad Spend (INR)", "sum"),
        Conversions=("Conversions", "sum")
    )
)

channel_cac["CAC (INR)"] = (
    channel_cac["Spend"] / channel_cac["Conversions"]
).round(2)

channel_cac.sort_values("CAC (INR)")
```

3. Data Analysis & Visualization

3.1. Overall Marketing Performance

```python
total_spend = df["Ad Spend (INR)"].sum()
total_leads = df["Leads"].sum()
total_conversions = df["Conversions"].sum()
total_revenue = df["Revenue (INR)"].sum()

overall_roi = (
    (total_revenue - total_spend)
    / total_spend * 100
)

overall_conversion_rate = (
    total_conversions / total_leads * 100
)

overall_cac = (
    total_spend / total_conversions
)

overall_cpl = (
    total_spend / total_leads
)

print("Total Ad Spend: ₹", round(total_spend, 2))
print("Total Leads:", total_leads)
print("Total Conversions:", total_conversions)
print("Total Revenue: ₹", round(total_revenue, 2))
print("Overall ROI: ", round(overall_roi, 2), "%")
print("Overall Conversion Rate: ", round(overall_conversion_rate, 2), "%")
print("Overall CPL: ₹", round(overall_cpl, 2))
print("Overall CAC: ₹", round(overall_cac, 2))
```

3.2. Which marketing channel performs best?

```python
channel_performance = (
    df.groupby("Channel")
    .agg(
        Spend=("Ad Spend (INR)", "sum"),
        Impressions=("Impressions", "sum"),
        Clicks=("Clicks", "sum"),
        Leads=("Leads", "sum"),
        Conversions=("Conversions", "sum"),
        Revenue=("Revenue (INR)", "sum")
    )
)

channel_performance["CTR (%)"] = (
    channel_performance["Clicks"]
    / channel_performance["Impressions"] * 100
)

channel_performance["CPC (INR)"] = (
    channel_performance["Spend"]
    / channel_performance["Clicks"]
)

channel_performance["CPL (INR)"] = (
    channel_performance["Spend"]
    / channel_performance["Leads"]
)

channel_performance["Conversion Rate (%)"] = (
    channel_performance["Conversions"]
    / channel_performance["Leads"] * 100
)

channel_performance["CAC (INR)"] = (
    channel_performance["Spend"]
    / channel_performance["Conversions"]
)

channel_performance["ROI (%)"] = (
    (channel_performance["Revenue"]
     - channel_performance["Spend"])
    / channel_performance["Spend"] * 100
)

channel_performance = channel_performance.round(2)

channel_performance.sort_values(
    "ROI (%)",
    ascending=False
)
```

3.3. Which is the best channel?

```python
best_roi_channel = channel_performance["ROI (%)"].idxmax()
best_cac_channel = channel_performance["CAC (INR)"].idxmin()
best_cpl_channel = channel_performance["CPL (INR)"].idxmin()
best_revenue_channel = channel_performance["Revenue"].idxmax()

print("Best ROI Channel:", best_roi_channel)
print("Best CAC Channel:", best_cac_channel)
print("Best CPL Channel:", best_cpl_channel)
print("Highest Revenue Channel:", best_revenue_channel)
```

3.4. Which campaigns are performing well and which need improvement?

```python
campaign_performance = (
    df.groupby("Campaign")
    .agg(
        Spend=("Ad Spend (INR)", "sum"),
        Impressions=("Impressions", "sum"),
        Clicks=("Clicks", "sum"),
        Leads=("Leads", "sum"),
        Conversions=("Conversions", "sum"),
        Revenue=("Revenue (INR)", "sum")
    )
)

campaign_performance["CTR (%)"] = (
    campaign_performance["Clicks"]
    / campaign_performance["Impressions"] * 100
)

campaign_performance["CPL (INR)"] = (
    campaign_performance["Spend"]
    / campaign_performance["Leads"]
)

campaign_performance["Conversion Rate (%)"] = (
    campaign_performance["Conversions"]
    / campaign_performance["Leads"] * 100
)

campaign_performance["CAC (INR)"] = (
    campaign_performance["Spend"]
    / campaign_performance["Conversions"]
)

campaign_performance["ROI (%)"] = (
    (campaign_performance["Revenue"]
     - campaign_performance["Spend"])
    / campaign_performance["Spend"] * 100
)

campaign_performance = campaign_performance.round(2)

campaign_performance.sort_values(
    "ROI (%)",
    ascending=False
)
```

3.5. Best & Worst Campaigns

```python
best_campaign = campaign_performance["ROI (%)"].idxmax()
worst_campaign = campaign_performance["ROI (%)"].idxmin()

print("Best Performing Campaign:", best_campaign)
print("Lowest Performing Campaign:", worst_campaign)
```

3.6. Monthly Performance Analysis

```python
df["Month"] = df["Date"].dt.to_period("M").astype(str)
```

```python
monthly_performance = (
    df.groupby("Month")
    .agg(
        Spend=("Ad Spend (INR)", "sum"),
        Leads=("Leads", "sum"),
        Conversions=("Conversions", "sum"),
        Revenue=("Revenue (INR)", "sum")
    )
)

monthly_performance["ROI (%)"] = (
    (monthly_performance["Revenue"]
     - monthly_performance["Spend"])
    / monthly_performance["Spend"] * 100
)

monthly_performance["Conversion Rate (%)"] = (
    monthly_performance["Conversions"]
    / monthly_performance["Leads"] * 100
)

monthly_performance["CAC (INR)"] = (
    monthly_performance["Spend"]
    / monthly_performance["Conversions"]
)

monthly_performance = monthly_performance.round(2)

monthly_performance
```

3.7. Best & Worst Months

```python
best_month = monthly_performance["ROI (%)"].idxmax()
worst_month = monthly_performance["ROI (%)"].idxmin()

print("Best Month:", best_month)
print("Best Month ROI:", monthly_performance.loc[best_month, "ROI (%)"], "%")

print("Worst Month:", worst_month)
print("Worst Month ROI:", monthly_performance.loc[worst_month, "ROI (%)"], "%")
```

3.8. Marketing Funnel Analysis

```python
funnel = {
    "Impressions": df["Impressions"].sum(),
    "Clicks": df["Clicks"].sum(),
    "Leads": df["Leads"].sum(),
    "Conversions": df["Conversions"].sum()
}

funnel
```

```python
impressions = funnel["Impressions"]
clicks = funnel["Clicks"]
leads = funnel["Leads"]
conversions = funnel["Conversions"]

print("Impression → Click Rate:",
      round(clicks / impressions * 100, 2), "%")

print("Click → Lead Rate:",
      round(leads / clicks * 100, 2), "%")

print("Lead → Conversion Rate:",
      round(conversions / leads * 100, 2), "%")
```
